```python
def repeat(s: str, n: int) -> str:
    return s * n

repeat("abc", 3)
```

```python
# Was passiert, wenn Sie repeat() mit zwei Zahlen aufrufen?
# Was passiert, wenn Sie repeat() mit zwei Strings aufrufen?

repeat(3, 3)
repeat("abc", "def")
```
