```python
# Teil 05 Part 06 - Workshop: for-Schleife
# Schreiben Sie eine Funktion print_all(items: list), die die Elemente der Liste items auf dem Bildschirm ausgibt,
# jeweils ein Element pro Zeile
# Was passiert, wenn Sie die Funktion mit einem String als Argument aufrufen, z.B. print_all("abc")
```

```python
items = [1, 2, 3]

def print_all(items: list):
    for item in items:
        print(item)

print_all(items)
print_all("abc")
```

```python
# Teil 05 Part 06 - Workshop: Einkaufsliste
# Definiren Sie Variablen:
# meine_einkaufsliste, die eine Liste mit den Strings Tee und Kaffee enthält
# eine_andere_einkaufsliste, die ebenfalls eine Liste mit den Strings Tee und Kaffee enthält
# Definieren Sie eine Funktion drucke_einkaufsliste(einkausliste), die die als Argument übergebene Einkaufsliste ausdruckt
# Testen Sie die Funktion drucke_einkaufsliste(einkaufsliste) mit beiden Einkaufslisten
# Definieren Sie eine Funktion kauf(produkt, einkaufsliste), das produkt zu einkaufsliste hinzufügt
```

```python
# 1. Variablen definieren
meine_einkaufsliste = ["Tee", "Kaffee"]
eine_andere_einkaufsliste = ["Tee", "Kaffee"]

# 2. Funktion definieren
def drucke_einkaufsliste(einkaufsliste):
    for artikel in einkaufsliste:
        print(artikel) # alternativ mit 'end = " "', um Artikel in einer zeiel darzustellen

# 3. Funktion mit beiden Listen testen
print("--- Test mit meine_einkaufsliste ---")
drucke_einkaufsliste(meine_einkaufsliste)
print("--- Test mit eine_andere_einkaufsliste ---")
drucke_einkaufsliste(eine_andere_einkaufsliste)

# 4. Erstellen Funktion Produkt hinzufügen
def kauf(produkt, einkaufsliste):
    einkaufsliste.append(produkt)

kauf("Milch", meine_einkaufsliste)
print("--- Test mit meine_einkaufsliste und weiteren Artikel ---")
drucke_einkaufsliste(meine_einkaufsliste)
#print(meine_einkaufsliste, end = " ")
```
