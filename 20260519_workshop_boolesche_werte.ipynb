```python
# Gegeben sind folgende Variablendefinitionen
# var1 = 2**2**4
# var2 = 3**12
# var3 = 100 * 200 * 300
# var4 = 4**3**2
# Ist das maximum von var1 und var2 größer als das Minimum von var3 und var4?
```

```python
var1 = 2**2**4
var2 = 3**12
var3 = 100 * 200 * 300
var4 = 4**3**2

ergebnis = max(var1, var2) > min(var3, var4)
print(ergebnis)
```

```python
# Teil 04 Part 06 - Workshop: Operatoren, Vergleiche
# var1 = 3 ** (3 * 4)
# var2 = 4**3**2
# var3 = (3**3) ** 3
# var4 = (2**3) ** 4
```

```python
# Ist var1 durch var3 teilbar und gleichzeitig var2 und durch var4 tailbar?
var1 = 3 ** (3 * 4) # 3**12 --> Basis 3
var2 = 4**3**2 # 4**9 = 2**2**9 = 2**18 --> Basis 2
var3 = (3**3) ** 3 # 3**9 --> Basis 3
var4 = (2**3) ** 4 # 2**12 --> Basis 2

ergebnis = (var1 % var3 == 0) and (var2 % var4 == 0) # Prüfung mit Modulo und 'and' Vergleichsoperator
print(ergebnis) 
```
