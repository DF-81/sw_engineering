```python
class Address:
    def __init__(self, street, number, zip_code, city):
        self.street = street
        self.number = number
        self.zip_code = zip_code
        self.city = city

    def __str__(self):
        return f"{self.street} {self.number}, {self.zip_code} {self.city}"

    def __repr__(self):
        return f"Address('{self.street}', {self.number}, '{self.zip_code}', '{self.city}')"
```

```python
class Project:
    def __init__(self, title):
        self.title = title
        self.requirements = []  # Initialize requirements list
        

    def add_requirement(self, skill):
        if not hasattr(self, 'requirements'):
            self.requirements = []
        self.requirements.append(skill)

    def __str__(self):
        return f"Project: {self.title}"

    def __repr__(self):
        return f"Projekt('{self.title}')"
```

```python
class Company:
    def __init__(self, name, address):
        self.name = name
        self.address = address  # has one address
        self.emp_list = []  # list for 1-N employees

    def add_employee(self, name, project=None):
        """ Company generate Employee and employee dosn´t need the company as a parametre. No more circular dependencies."""
        new_employee = Employee(name, project) # Employee is not defined!?
        self.emp_list.append(new_employee)
        return new_employee

    def remove_employee(self, employee):
        """ Remove employee from emp_list """
        if employee in self.emp_list:
            self.emp_list.remove(employee)

    def __str__(self):
        return f"Company {self.name} (Sitz: {self.address})"

    def __repr__(self):
        return f"Company(name='{self.name}', adress={repr(self.address)})"
```

```python
class Employee:
    def __init__(self, name, project=None):
        self.name = name
        self.project = project  # is working at a project
        self.addresses_list = []  # can have 1 to N addresses
        self.addresses_dict = {}  # Alternative: use a dictionary to store addresses with labels (e.g., 'home', 'work')
        self.skills_list = [] # Add skills attribute to Employee class
    
    def add_address(self, address): # use list for address 
        self.addresses_list.append(address)
    
    def add_address_with_label(self, label, address): # use dictionary for address
        self.addresses_dict[label] = address
    
    def add_skills(self, skills):
        if not hasattr(self, 'skills'):
            self.skills = []
        self.skills_list.append(skills)

    def __str__(self):
        project_info = self.project.title if self.project else "No Project"
        return f"Employee {self.name} Project: {project_info})"

    def __repr__(self):
        return f"Employee(name='{self.name}'"
```

```python
class Skills:
    def __init__(self, skill_name, level):
        self.skill_name = skill_name
        self.level = level

    def __str__(self):
        return f"{self.skill_name} (Level: {self.level})"

    def __repr__(self):
        return f"Skills(skill_name='{self.skill_name}', level='{self.level}')" 
     
    def check_matching(employee, project):
        if not hasattr(project, 'requirements'):
            return True  # No requirements means everyone matches
        if not hasattr(employee, 'skills'):
            return False  # Employee has no skills, can't match any requirements

        employee_skill_names = {skill.skill_name for skill in employee.skills}
        for requirement in project.requirements:
            if requirement.skill_name not in employee_skill_names:
                return False  # Missing required skill
        return True  # All requirements are met
```

```python

# 1. Build Infrastructure
headquarter = Address("Firmengasse", 1, 10115, "Berlin")
my_company = Company("TechCorp", headquarter)

# 2. Create Project
ki_project = Project("AI-Assistent")

# 3. Create Employee (linked to Company and Project)
max = my_company.add_employee("Max Mustermann", ki_project)

# 4. Add Employee Addresses (1-N)
# Option 1: Add addresses without labels
privat = Address("Privatweg", "5", 12345, "Musterhausen")
ferienhaus = Address("Strandallee", "7", 23552, "Lübeck")
max.add_address(privat)
max.add_address(ferienhaus)

# Alternative Option 2: Add addresses with labels
private_address = Address("Privatweg", "5", 12345, "Musterhausen")
work_address = Address("TechCorp Street", "1", 10115, "Berlin")
max.add_address_with_label("home", private_address)
max.add_address_with_label("work", work_address)

# 5. Add Skills to project and employee
python_skill = Skills("Python", "Advanced")
sql_skill = Skills("SQL", "Medior")
ki_project.add_requirement(python_skill)
ki_project.add_requirement(sql_skill)
max.add_skills(python_skill)
max.add_skills(sql_skill)

# --- Print out ---
print(f"Mitarbeiter bei {my_company.name}: {[m.name for m in my_company.emp_list]}")
print(max)

# use addres with list
print(f"Count addresses from Max: {len(max.addresses_list)}")
print(f"Addresses from Max: {max.addresses_list}")

# use address with dict
print(f"Addresses with labels from Max: {max.addresses_dict}")

print(f"Employees in the Company: {[m.name for m in my_company.emp_list]}")
print(f"Projekt-Anforderung: {ki_project.requirements}")
print(f"Max besitzt: {max.skills_list}")
```
