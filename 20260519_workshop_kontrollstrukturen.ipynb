```python
# Teil 05 Part 04 - Workshop: Gerade zahl, assertion, verschachtelte Funktion
# Schreiben Sie eine Funktion ist_gerade(zahl), die True zurückgibt,
# falls zahl gerade ist und False, falls zahl ungerade ist
# Schreiben Sie Assertions, die testen ob ist_gerade() für die Argumente 0,1 und 8 das korrekte Ergebnis liefert.
# Schreiben Sie eien Funktion drucke_ist_gerade(zahl), die:
#   - {zahl} ist gerade. auf dem Bildschirm ausgibt, falls zahl gerade ist und
#   - {zahl} ist ungerade. auf dem Bildschirm ausgibt, falls zahl nicht gerade ist
# Verwenden Sie dabei die Funktion ist_gerade(), um zu überprüfen, ob zahl gerade ist
# Testen Sie drucke_ist_gerade(zahl) mit den Werten 0,1 und 8
```

```python
def ist_gerade(zahl):
    return zahl % 2 == 0
ist_gerade(7)
```

```python
assert ist_gerade(0)
assert ist_gerade(1), "Zahl sollte gerade sein"
assert ist_gerade(8)
```

```python
def drucke_ist_gerade(zahl):
    if ist_gerade(zahl):
        print(f"{zahl} ist gerade")
    else:
        print(f"{zahl} ist ungerade")

drucke_ist_gerade(0)
drucke_ist_gerade(1)
drucke_ist_gerade(8)
```

```python
# Teil 05 Part 04 - Workshop: Fits in line
# Schreiben Sie eine Funktion fits_in_line(text: str, line_length: int = 72),
# die True oder False zurückgivbt, je nachdem ob text in einer Zeile der Länge line_length ausgegeben werden kann oder nicht
# Hinweis: Sie können die Länge eines Strings mit der Funktion len() bestimmen
```

```python
def fits_in_line(text: str, line_length: int = 72) -> bool:
    return len(text) <= line_length
    
print(fits_in_line("Hallo", 3))
print(fits_in_line("Hallo", 10))
```
