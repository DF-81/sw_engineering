```python
def effektive_spende(EURO):
    spende_verdoppeln = EURO * 2
    spende_gesamt = spende_verdoppeln + 10
    return spende_gesamt # hier wird der Wert innerhalb der Funktion zurückgegeben

print(effektive_spende(10))
print(effektive_spende(25))
print(effektive_spende(50))
print(effektive_spende(100))
print(effektive_spende(500))   

spende_effektiv = effektive_spende(100) # Wert wird außerhalb in einer neuen Variable gespeichert
print(spende_effektiv)
print(type(spende_effektiv)) # class 'int' oder class 'float'
```

```python
# Schreiben Sie eine Funktion drucke_ist_gerade(zahl), die
# {zahl} ist gerade. auf dem Bildschirm ausgibt, falls zahl gerade ist und
# {zahl} ist ungerade. auf dem Bildschirm ausgibt, falls zahl nciht gerade ist.
# Verwenden Sie dabei die Dunktion ist_gerade(), um zu überprüfen, ob zahl gerade ist
# Testen Sie drucke_ist_gerade(zahl) mit den Werten 0, 1 und 8
```

```python
# Hilfsfunkion wie bereits zuvor geschrieben, die True oder False zurückgibt
def ist_gerade(zahl):
    return zahl % 2 == 0

# Hauptfunktion, die auf Bildschirm ausgegeben wird
def drucke_ist_gerade(zahl):
    if ist_gerade(zahl): # Hilfsfunktion als verschachtelte Funktion
        print(f"{zahl} ist gerade.")
    else:
        print(f"{zahl} ist ungerade.")

drucke_ist_gerade(0)
drucke_ist_gerade(1)
drucke_ist_gerade(8)
```

```python
# Teil Part 04 - Workshop: Verkürzte Ausgabe
# Schreiben Sie eine Funktion fits_in_line(text: str, line_length: int), die
# True oder False zurückgibt, je nachdem ob Text in einer Zeile der Länge line_length
# ausgegeben werden kann kann oder nicht
# Hinweis: Sie können die Länge eines Strings mit der Funktion len() bestimmen
```

```python
def fits_in_line(text: str, line_length: int) -> bool:
    """
    Prüft, ob ein Text in eine Zeile der Länge line_length passt.
    """
    return len(text) <= line_length

fits_in_line("Hallo", 4)
```
