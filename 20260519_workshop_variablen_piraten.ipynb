```python
# Teil 04 Part 4 - Workshop: Piraten
# Der Legende nach wurde die Beute bei piratenbanden gerecht durch alle Piraten geteilt.
# Falls sich die beute nicht gerecht aufteilen ließ, erhielt der kapitän den überschüssigen Anteil.
# Wie viele Golddublonen erhält jeder Pirat einer 8-köpfigen bande (7 piraten + 1 Kapitä),
# wenn ein Schatz mit 1000 Golddublonen erbeutet wurde? (Verwenden Sie Variablen, um die Berechnung klarer zu machen)  
```

```python
# Variablen definieren
anzahl_piraten = 7
schatz_gesamt = 1000

# Berechnung des gleichmäßigen Anteils (Ganzzahl-Division)
# Der Operator // schneidet die Nachkommastellen einfach ab
anteil_pro_person = schatz_gesamt // (anzahl_piraten + 1)

# Berechnung des Rests für den Kapitän (Modulo-Operator)
# % gibt den Rest einer Division zurück
rest_schatz = schatz_gesamt % (anzahl_piraten + 1)

# Der Kapitän bekommt seinen normalen Anteil plus den Rest
anteil_kapitaen = anteil_pro_person + rest_schatz
anteil_kapitaen_extra = anteil_kapitaen - anteil_pro_person

# Ergebnisse ausgeben

print(f"Es sind {anzahl_piraten} Piraten an Bord.")
print(f"Jeder Pirat erhält: {anteil_pro_person} Goldstücke")
print(f"Der Kapitän erhält: {anteil_kapitaen} Goldstücke und somit {anteil_kapitaen_extra} Goldstücke extra")
```

```python
# Die Piratenbande nimmt 3 neue Piraten-Lehrlinge auf.
# Wie verändert sich die Aufteilung der Beute?
```

```python
# Variablen definieren
anzahl_piraten = 7
anzahl_piraten = anzahl_piraten + 3 # Es kommen 3 neue Piraten dazu, also erhöhen wir die Anzahl der Piraten um 3
schatz_gesamt = 1000

# Berechnung des gleichmäßigen Anteils (Ganzzahl-Division)
# Der Operator // schneidet die Nachkommastellen einfach ab
anteil_pro_person = schatz_gesamt // (anzahl_piraten + 1)

# Berechnung des Rests für den Kapitän (Modulo-Operator)
# % gibt den Rest einer Division zurück
rest_schatz = schatz_gesamt % (anzahl_piraten + 1)

# Der Kapitän bekommt seinen normalen Anteil plus den Rest
anteil_kapitaen = anteil_pro_person + rest_schatz
anteil_kapitaen_extra = anteil_kapitaen - anteil_pro_person

# Ergebnisse ausgeben

print(f"Es sind jetzt {anzahl_piraten} Piraten an Bord.")
print(f"Jeder Pirat erhält: {anteil_pro_person} Goldstücke")
print(f"Der Kapitän erhält: {anteil_kapitaen} Goldstücke  und somit {anteil_kapitaen_extra} Goldstücke extra")
```

```python
# Wie könnte eine Funktion aussehen, um die Beute in verschiedenen Szenarien darzustellen?
```

```python
def berechne_beute(piraten_anzahl, schatz_betrag):
    # Gesamtzahl der Köpfe (Crew + Kapitän)
    personen_gesamt = piraten_anzahl + 1
    
    # Berechnung
    anteil_pirat = schatz_betrag // personen_gesamt
    rest_schatz = schatz_betrag % personen_gesamt
    anteil_kapitaen = anteil_pirat + rest_schatz
    anteil_kapitaen_extra = anteil_kapitaen - anteil_pirat

    return anteil_pirat, anteil_kapitaen, anteil_kapitaen_extra

# --- Testlauf mit verschiedenen Szenarien ---

# Szenario 1: Die ursprünglichen 7 Piraten
p7_anteil, k7_anteil, k7_extra = berechne_beute(7, 1000)

# Szenario 2: Die 10 Piraten (nach der Verstärkung)
p10_anteil, k10_anteil, k10_extra = berechne_beute(10, 1000)

print(f"Bei 7 Piraten: Pirat {p7_anteil}, Kapitän {k7_anteil} und somit {k7_extra} Goldstücke extra")
print(f"Bei 10 Piraten: Pirat {p10_anteil}, Kapitän {k10_anteil} und somit {k10_extra} Goldstücke extra")
```
