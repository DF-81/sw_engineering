```python
# Teil 9 Part 5 - Proberties - Workshop: Buchinformation
# Erstellen Sie eine Klasse Book, die Informationen über ein Buch verwaltet
# Implementieren Sie Titel und Autor als schreibgeschützte Attribute
# Implementieren Sie eine Property citation, die einen Zitationsstring im Format "Titel" by Auto zurückgibt
# Fügen Sie eine __repr__ Methode hinzu, die Titel und Autor anzeigt
# Testen Sie die Klasse, indem Sie ein Buchobjekt erstellen und die citation-Property abrufen 
```

```python
class Book:
    def __init__(self, titel, autor):
        self._titel = titel # underline indicates that these attributes are intended to be private
        self._autor = autor # underline indicates that these attributes are intended to be private
    
    @property
    def citation(self):
        return f'"{self._titel}" by {self._autor}'

    def __repr__(self):
        return f'Book(titel="{self._titel}", autor="{self._autor}")'

# Test 
buch = Book("Robinson Crusoe", "Daniel Defoe")
print(buch.citation)  # Ausgabe: "Robinson Crusoe" by Daniel Defoe
print(buch)           # Ausgabe: Book(titel="Robinson Crusoe", autor="Daniel Defoe")
```

```python
# Teil 9 Part 6 - Proberties und Setter - Workshop: Temperatur-Konverter
# Erstellen Sie eine Klasse Temperature, die Temperaturen intern in Grad Celsius speichert (_celsius)
# Implementieren Sie Properties celsius, fahrenheit und kelvin
# Der Konstruktor soll einen Wert und eine Einheit ('C', 'F', 'K') entgegennehmen und den internen _celsius-Wert entsprechend setzen
# Das Setzen einer der Properties (celsius, fahrenheit, kelvin) soll den internen _celsius-Wert aktualisieren
# Das Abrufen jeder Property soll den korrekt umgerechneten Wert zurückgeben
# Fügen Sie eine __repr__ Methode hinzu, die die Temperatur in allen drei Einheiten anzeigt
# Formeln: C = (F-32) * 5/9, F = C * 9/5 + 32, C = K - 273.15, K = C + 273.15
```

```python
class Temperature: # create class Temperature
    
    def __init__(self, value, unit="C"): # here comes the contructor __init__
        if unit == 'C':
            self.celsius = value # call the Celsius-Setter
        elif unit == 'F':
            self.celsius = value # call the Fahrenheit-Setter 
        elif unit == 'K':
            self.celsius = value # call the Kelvin-Setter
        else:
            raise ValueError("Unit must be 'C', 'F', or 'K'") # if unit is none of the accepted raise an error

    # Celsius Property
    # Getter
    @property
    def celsius(self):
        return self._celsius # save C for temperature as _.celsius
    
    # Setter
    @celsius.setter
    def celsius(self, value):
            if value < -273.15: # additional feature to avoid temperatures lower than absolute zero  
                raise ValueError("Temperature can´t be lower than -273.15")
            self._celsius = float(value)

    # Fahrenheit Property
    # Getter
    @property
    def fahrenheit(self):
        return self._celsius * 9 / 5 + 32 # given F in C for temperature as _.celsius to save
    
    # Setter
    @fahrenheit.setter
    def fahrenheit(self, value):
        self.celsius = (float(value - 32) * 5 / 9) # with new feature the self._celsius has changed to self.celsius-Setter

    # Kelvin Property
    # Getter
    @property
    def kelvin(self):
        return self._celsius + 273.15 # given K in C for temperature as _.celsius to save
    
    # Setter
    @kelvin.setter
    def kelvin(self, value):
        self.celsius = float(value - 273.15) # with new feature the self._celsius has changed to self.celsius-Setter

    # representation
    def __repr__(self):
        return f"Temperature(celsius={self.celsius:.2f}°C, fahrenheit={self.fahrenheit:.2f}°F, kelvin={self.kelvin:.2f}K)"

# 1. Create some instances and show the output
t1 = Temperature(50, 'C')
t2 = Temperature(32, 'F')
t3 = Temperature(25, 'K')
print(t1)
print(t2)
print(t3)

# 2. print specific value
print(t1.fahrenheit)

# 3. set new value and show the output
t3.kelvin = 50
print(t3.kelvin)
print(t3)
```

```python
# Teil 9 Part 6 - Proberties und Setter - Workshop: Rechteck-Geometrie
# Erstellen Sie eine Klasse Rectangle, die width (Breite) und height (Höhe) als private Attribut (_width, _height) speichert
# Stellen Sie Validierung hinzu, um sicherzustellen, dass Breite und Höhe nicht negativ gesetzt werden können
# Implementieren Sie nur lesbare Proberties area (Fläche) und perimeter (Umfang), die aus _width und _height berechnet werden
# Fügen Sie eine __repr__-Methode hinzu, die Breite, Höhe, Fläche und Umfang anzeigt
# Formeln: Fläche = Breite * Höhe, Umfang = 2 * (Breite+Höhe)
```

```python
class Rectangle:
    def __init__(self, width, height):
        self.width = width
        self.height = height
    
    # Getter and Setter für width
    @property
    def width(self):
        return self._width
    
    @width.setter
    def width(self, value):
            if value < 0: # additional feature to avoid temperatures lower than absolute zero  
                raise ValueError("Dimension width can´t be lower than 0")
            self._width = float(value)

    # Getter and Setter für height
    @property
    def height(self):
         return self._height
    
    @height.setter
    def height(self, value):
        if value < 0:
             raise ValueError("Dimension height can´t be lower than 0")
        self._height = float(value)         

    # Import the valitade dimensions ._width and ._height for area and perimeter 
    @property
    def area(self):
        return self.width * self.height
    
    @property
    def perimeter(self):
         return 2 * (self.width + self.height)
    
    def __repr__(self):
        return (f"Rectangle(Breite: {self.width:.2f}, Höhe: {self.height:.2f}, "
               f"Fläche: {self.area:.2f}, Umfang: {self.perimeter:.2f})")
    
r1 = Rectangle(1, 5)
print(r1)

r1.width = 4
print(r1)
```
