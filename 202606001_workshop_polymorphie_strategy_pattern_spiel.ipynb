```python
# 1. Define Interface
from abc import ABC, abstractmethod
class FightStrategy(ABC):
    @abstractmethod
    def fight(self):
        #print("Let´s fight!")
    # usually we use pass and would not implement the method in the interface, but for demonstration purposes we do it here.
    # The method can be overridden in the subclasses.
    # pass or another workaroud if the method is not implemented from the beginning of the interface/game.
        if self.fight_strategy is None:
            print(f"{self.name} has no weapon!")
        else:
            print(f"{self.name} is fighting!")
            self.fight_strategy.fight()
```

```python
# 2. Create Class Fight with Sword as strategy
class FightWithSword(FightStrategy):
    def fight(self):
        print("Strategy: Fight with sword!")
```

```python
# 3. Create Class Fight with Magic as strategy
class FightWithMagic(FightStrategy):
    def fight(self):
        print("Strategy: Fight with magic!")
```

```python
# 4. Create Class Fight with Gun as strategy
class FightWithGun(FightStrategy):
    def fight(self):
        print("Strategy: Fight with gun!")
```

```python
# 5. Create Avatar Class
class Avatar:
    def __init__(self, name=None, fight_strategy=None):
        self.name = name
        self.fight_strategy = fight_strategy
    
    def set_fight_strategy(self, strategy):
        self.fight_strategy = strategy

    def fight(self):
        print(f"{self.name} is fighting!")
        self.fight_strategy.fight() # Here the magic happens. The fight method of the strategy is called, which can be different for each avatar. fight_strategy is an instance of the strategy class, which can be changed at runtime but fight never changes. This is the essence of the strategy pattern.
```

```python
# 6. Create Main Logic

if __name__ == "__main__":
    # Create Avatar as constructor parameter for __init__
    witcher = Avatar("The Witcher")

    # use Setter method to change the strategy at runtime
    # 1. Fight with Sword
    witcher.set_fight_strategy(FightWithSword())
    witcher.fight()

    # 2. Fight with Magic
    witcher.set_fight_strategy(FightWithMagic())
    witcher.fight()

    # 3. Fight with Gun
    witcher.set_fight_strategy(FightWithGun())
    witcher.fight()

    # Create a second Avatar and set the strategy at runtime
    Danyo = Avatar("The Barbarian")
    Danyo.set_fight_strategy(FightWithSword())
    Danyo.fight()
```
